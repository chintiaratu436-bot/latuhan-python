```
import numpy as np

suhu = np.array([25.1, 25.8, 26.4, 27.2, 27.9, 28.5, 29.1, 29.8, 30.2, 30.8, 31.5])
```

```
print ("data temperature:", suhu)

```

```
print ("jumlah:", suhu.size)

print ("dimensi:", suhu.ndim)
```

```
print ("nilai minimum:", np.min(suhu))

print("nilai maksimum:", np.max(suhu))

print("rata rata:", np.mean(suhu))

print("jumlah:", np.sum (suhu))

print("median:", np.median(suhu))

print("standar deviasi:", np.std(suhu))
```

```
suhu_di_atas_29 = suhu [suhu > 29]

print("Temperatur di atas 29 °C:", suhu_di_atas_29)
```

```
fahrenheit = suhu * 9 / 5 + 32

print("Mengubah Celsius menjadi Fahrenheit:", fahrenheit)
```
